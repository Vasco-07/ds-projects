```python
import numpy as np
import pandas as pd
data = {'A': [1, 2, np.nan],
        'B': [4, np.nan, np.nan]}
df = pd.DataFrame(data)
print(df.isna())
df_clean = df.dropna()
df_filled = df.fillna(0)
print(df_clean)
print(df_filled)
df = df.drop_duplicates()
```

```python
import pandas as pd
df = pd.DataFrame({
    'gender': ['male', 'female', 'male']
})
df['gender_num'] = df['gender'].map({
    'male': 0,
    'female': 1
})
df_renamed = df.rename(columns={'gender': 'Gender'})
print(df_renamed)
```
